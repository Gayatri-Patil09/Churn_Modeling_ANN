#                          Part ( A )

### 1. Data Preprocessing
### Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### 2. Load Dataset

```python
dataset=pd.read_csv("data/Churn_Modelling.csv")
#  View Dataset
print(dataset.head(2))
print(dataset.columns)
```

### 3. Check for important columns which affect our analysis

```python
# saperate dependent and independent value
X= dataset.iloc[:,3:13].values  # first val include and last not include so use 13 , i.e we can get 12 no. values
y= dataset.iloc[:,13].values
```

### 4. Encoding of Categorical data
        In our dataset Geography and Gender attribute is in text format and we enocode these two attribute in this part by using LabelEncoder and OneHotEncoder from sklearn.preprocessing library

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer

labelencoder_X1= LabelEncoder()
labelencoder_X1.fit_transform(X[:,1])

labelencoder_X2= LabelEncoder()
labelencoder_X2.fit_transform(X[:,2])

# identify indices or name of categorical col
categorical_cols= [1,2] # replace with col indices

#create columnTransformer
ct= ColumnTransformer(transformers=[ ('encoder', OneHotEncoder(), categorical_cols )],
                      remainder= 'passthrough') #keep all other col as they are )

# apply it to your dataset
X= ct.fit_transform(X)

#onehotencoder= OneHotEncoder(categorical_feature = [1])
# X = onehotencoder.fit_transorm(X).toarry()
X= X[:,1:]
X
```

### 5. Split data for Train (80%), Test(20)

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test= train_test_split(X,y,test_size=0.2, random_state=0)
```

### 6. Feature Scalling
        It is most important part of data preprocessing. our datset, some attribute contains info in num val, some very high and some very low. If we see age, estimated salary. this will cause some issue in machinery model to solve that problem we set all val on same scale there ae two methods to solve this issue. 1. Normalization 2. Standard Scaler

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train= sc.fit_transform(X_train)
X_test=sc.transform(X_test)
            
```

#  Part ( B )

## 1. Building our model - ANN  model
###  we import lib which will build our ANN model - keras which build deep Neural network based on tesnsorflow, because we use tensorflow backhand. Here we import two module from kers
 ### i. sequential use for initilizing our ANN model
 ### ii. dense use for adding different layer of ANN

```python
import tensorflow.keras
# from sklearn.models import sequential
# from sklearn.layers import Dense

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# initilize our ANN model using sequential model
classifier = Sequential()
```

```python
# add input layer and first hidden layer 
classifier.add(Dense(units=1,activation='relu'))
classifier.add(Dense(1,activation='relu'))
```

## 2. Fitting ANN

```python
# 1. Check data shapes
print("--- Data Shapes ---")
print(f"X_train shape: {X_train.shape} | Data type: {type(X_train)}")
print(f"y_train shape: {y_train.shape} | Data type: {type(y_train)}")

# 2. Check model structure
print("\n--- Model Architecture ---")
classifier.summary()
```

## 3. Compile ANN


```python
#classifier.compile(optimizer='adam',loss='binary-entropy', metrics=['accuracy'])

classifier.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
import tensorflow as  tf
classifier = tf.keras.Sequential([
    # Notice the comma: (12,) means 12 features, any batch size
    tf.keras.layers.Dense(units=6, activation='relu', input_shape=(12,)), 
    tf.keras.layers.Dense(units=6, activation='relu'),
    tf.keras.layers.Dense(units=1, activation='sigmoid')
])

```

```python
classifier.fit(X_train,y_train,batch_size=10,epochs=100)
```

# PART (C)
### Make Prediction and accuracy result

```python
# predict test set result
y_pred=classifier.predict(X_test)
y_pred=(y_pred > 0.5)
y_pred
```

```python
# 2. Confusion metrics
#     make CM for our test result set . y_test ( actual result) , y_pred(predicted result)
from sklearn.metrics import confusion_matrix
confusion_matrix= confusion_matrix(y_test, y_pred)
confusion_matrix
```

```python
# 3. accuracy score
from sklearn.metrics import accuracy_score
accuracy= accuracy_score(y_test,y_pred)
accuracy
```
